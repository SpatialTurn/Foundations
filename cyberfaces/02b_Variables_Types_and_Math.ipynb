## Variables

A **variable** is a name that points to a value. You create one with `=`, which means "store the value on
the right under the name on the left" (not "is equal to").

````python
county = "Tippecanoe"
population = 177513
land_sq_mi = 499.8
print(county, population, land_sq_mi)
````

Rules and habits for names:

- use letters, numbers and underscores, and do not start with a number (`pop_2012`, not `2012pop`)
- no spaces: `land_sq_mi`, not `land sq mi`
- names are case sensitive: `County` and `county` are different
- choose names that say what the value is: `population` is better than `p` or `x`

A variable keeps its value until you give it a new one:

````python
year = 2012
print(year)
year = 2024        # replace the value
print(year)
year = year + 1    # use the old value to make a new one
print(year)
year += 1          # a shortcut for year = year + 1
print(year)
````

## Data types

Every value has a **type**. These four cover most of what you will meet:

| Type | Name in Python | Examples |
|---|---|---|
| Whole number | `int` | `177513`, `-3` |
| Decimal number | `float` | `499.8`, `0.5` |
| Text (string) | `str` | `"Tippecanoe"`, `'18157'` |
| True or false | `bool` | `True`, `False` |

`type()` tells you the type of any value.

````python
print(type(population))
print(type(land_sq_mi))
print(type(county))
print(type(True))
````

## Math

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `+` `-` | add, subtract | `10 - 3` | `7` |
| `*` | multiply | `10 * 3` | `30` |
| `/` | divide | `10 / 4` | `2.5` |
| `//` | divide and round down | `10 // 4` | `2` |
| `%` | remainder | `10 % 4` | `2` |
| `**` | power | `10 ** 2` | `100` |

Python follows the usual order of operations; use brackets to be clear.

````python
density = population / land_sq_mi        # people per square mile
print(density)
print(round(density, 1))                 # round to 1 decimal place
print((2 + 3) * 4, "versus", 2 + 3 * 4)
````

## Converting types

Numbers stored as text cannot be used in math. This matters a lot with real data: the Census API, for
example, returns every number as text. Use `int()`, `float()` and `str()` to convert.

````python
pop_text = "177513"           # looks like a number, but it is text
print(pop_text + pop_text)    # text + text joins the two strings
pop_number = int(pop_text)
print(pop_number + pop_number)
print("Population: " + str(pop_number))   # numbers must become text before joining to text
````

Mixing them without converting gives a `TypeError`:

````python
"Population: " + pop_number
````
