**GeoPandas** adds geography to pandas. A **GeoDataFrame** is a DataFrame with one extra column,
`geometry`, holding a shape for each row: a point, a line or a polygon. Everything you learned in pandas
(selecting, filtering, sorting) still works. GeoPandas comes installed in the SpatialTurn GIS kernel.

````python
import geopandas as gpd
import matplotlib.pyplot as plt
````

## Read polygons

`gpd.read_file()` reads GeoJSON, shapefiles (also zipped) and GeoPackages, straight from a web address.

````python
counties = gpd.read_file("https://raw.githubusercontent.com/SpatialTurn/Foundations/main/data/indiana_counties.geojson")
counties.head()
````

````python
print(type(counties))
print(list(counties.geom_type.unique()))     # what kind of shapes
print(counties.crs)                    # the coordinate reference system (notebook 04b)
````

## Plot the shapes

````python
counties.plot(edgecolor="white", figsize=(5, 7))
plt.title("Indiana counties")
plt.show()
````

## A choropleth map

A **choropleth** colors each area by a value. Pass the column name to `.plot()`, choose a color ramp with
`cmap`, and add a legend. Map rates and densities rather than raw counts, so that large counties do not
dominate just because they are large.

````python
ax = counties.plot(column="unemp_2016", cmap="YlOrRd", legend=True,
                   edgecolor="white", linewidth=0.5, figsize=(5, 7),
                   legend_kwds={"label": "Unemployment rate, 2016 (%)"})
ax.set_title("Unemployment by county, 2016")
ax.set_axis_off()       # latitude and longitude ticks add little here
plt.show()
````

## Filter, then map

Filtering works exactly as in pandas.

````python
tippecanoe = counties[counties["name"] == "Tippecanoe"]
tippecanoe.plot(color="lightgray", edgecolor="black")
plt.title("Tippecanoe County")
plt.show()
````

## Read points and layer them

The Case Studies module has a file of 64 food stores in Lafayette and West Lafayette, from OpenStreetMap.
To draw several layers on one map, draw the first layer, keep the axes it returns (`ax`), and pass
`ax=ax` to the next layer.

````python
stores = gpd.read_file("https://raw.githubusercontent.com/SpatialTurn/CaseStudies/main/data/grocery_stores.geojson")
print(list(stores.geom_type.unique()), len(stores), "stores")
stores[["store_name", "store_type", "city"]].head()
````

````python
ax = tippecanoe.plot(color="whitesmoke", edgecolor="black", figsize=(6, 6))
stores.plot(ax=ax, column="store_type", categorical=True, legend=True, markersize=15)
ax.set_title("Food stores around Lafayette and West Lafayette")
ax.set_axis_off()
plt.show()
````

Each point's `geometry` is a coordinate pair. You can get the coordinates as ordinary columns:

````python
stores["lon"] = stores.geometry.x
stores["lat"] = stores.geometry.y
stores[["store_name", "lon", "lat"]].head(3)
````

Saving works much like pandas: `counties.to_file("counties.gpkg", driver="GPKG")` writes a GeoPackage you can
open in QGIS.
