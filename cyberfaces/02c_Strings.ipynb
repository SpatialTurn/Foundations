## Making strings

A **string** is text inside quotes. Single and double quotes both work; pick one and be consistent.

````python
city = "West Lafayette"
state = 'Indiana'
place = city + ", " + state      # + joins strings
print(place)
````

## f-strings: the easy way to build text

Put an `f` before the opening quote, and anything inside `{ }` is replaced by its value. You can also
control how numbers look.

````python
population = 177513
density = 355.17
print(f"{place} is in a county of {population} people.")
print(f"With separators: {population:,}")          # 177,513
print(f"One decimal place: {density:.1f}")         # 355.2
````

## String methods

Strings come with built-in tools called **methods**. You call a method with a dot after the value. They are
very useful for cleaning messy data, such as names typed by different people.

| Method | What it does | Example result |
|---|---|---|
| `.upper()`, `.lower()` | Change case | `"INDIANA"` |
| `.title()` | Capitalize each word | `"West Lafayette"` |
| `.strip()` | Remove spaces at the start and end | `"Tippecanoe"` |
| `.replace(a, b)` | Replace one piece of text with another | `"Tippecanoe Co."` |
| `.split(sep)` | Cut text into a list of parts | `["40.42", "-86.91"]` |

````python
messy = "   tippecanoe county  "
clean = messy.strip().title()        # methods can be chained
print(f"[{messy}] becomes [{clean}]")
print(clean.replace("County", "Co."))
print(len(clean))                    # len() counts characters
````

````python
coords = "40.4247,-86.9107"
parts = coords.split(",")            # a list of two strings
print(parts)
lat = float(parts[0])                # convert each part to a number
lon = float(parts[1])
print(lat, lon)
````

## Checking text

`in` tells you whether one piece of text appears inside another, and `==` checks whether two strings are
exactly the same. Comparisons are case sensitive, which is one more reason to clean text first.

````python
print("Lafayette" in "West Lafayette, Indiana")
print("indiana" == "Indiana")
print("indiana" == "Indiana".lower())
````
