## What is a notebook?

A **notebook** mixes two kinds of cells:

- **Text cells** (like this one) explain what is going on. They are written in Markdown.
- **Code cells** hold Python code. When you run a code cell, its output appears just below it.

Click the code cell below and press **Shift + Enter** to run it.

````python
print("Hello, World!")
````

`print()` is a **function**: a named piece of code that does a job. Whatever you put between the brackets is
shown as output. Text must be inside quotes; numbers do not need quotes.

````python
print("Welcome to Python")
print(2026)
print("The answer is", 6 * 7)     # several things, separated by commas
````

### The last line is shown automatically

If the last line of a cell is a value or a calculation, the notebook shows it even without `print()`.

````python
2 + 2
````

### Run cells in order

The number that appears next to a cell after you run it (for example `[3]`) shows the order in which
cells were run. Later cells often depend on earlier ones, so run them from top to bottom. If things get
confusing, use **Runtime > Restart session and run all** to start fresh.

## Comments

Anything after a `#` on a line is a **comment**. Python ignores it; it is a note for people reading the
code, including your future self. Good comments explain *why* you did something.

````python
# Convert square miles to square kilometers
sq_mi = 10
print(sq_mi * 2.59)   # 1 square mile is about 2.59 square kilometers
````
