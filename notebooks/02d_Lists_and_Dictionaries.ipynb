## Lists

A **list** holds several values in order, inside square brackets.

````python
counties = ["Marion", "Lake", "Allen", "Hamilton", "St. Joseph"]
populations = [918977, 493618, 360412, 289495, 266344]
print(counties)
print(len(counties), "counties")
````

### Picking items: indexes start at 0

Each item has a position, its **index**. Python counts from **0**, so the first item is `[0]`. Negative
numbers count from the end: `[-1]` is the last item.

````python
print(counties[0])     # first
print(counties[1])     # second
print(counties[-1])    # last
````

### Slices: several items at once

`list[start:end]` gives the items from `start` up to, but **not including**, `end`.

````python
print(counties[0:3])   # the first three: indexes 0, 1 and 2
print(counties[:2])    # from the start
print(counties[3:])    # to the end
````

### Changing a list and summarizing it

````python
counties.append("Tippecanoe")     # add to the end
populations.append(177513)
print(counties)
print("Total:", sum(populations))
print("Largest:", max(populations), "Smallest:", min(populations))
print("Sorted:", sorted(populations))
print("Is Lake in the list?", "Lake" in counties)
````

## Dictionaries

A **dictionary** stores **key: value** pairs inside curly brackets. You look values up by their key
instead of by position, which makes the code easier to read.

````python
tippecanoe = {
    "name": "Tippecanoe",
    "fips": "18157",
    "population": 177513,
    "land_sq_mi": 499.8,
}
print(tippecanoe["name"])
print(tippecanoe["population"] / tippecanoe["land_sq_mi"])
````

````python
tippecanoe["unemployment"] = 4.0          # add a new key
tippecanoe["population"] = 180000         # update an existing key
print(tippecanoe)
print(tippecanoe.keys())
print(tippecanoe.get("seat", "not recorded"))   # .get() gives a default instead of an error
````

The FIPS code `"18157"` is stored as **text**, not a number. Identifiers like FIPS codes, GEOIDs and ZIP
codes are labels, not quantities, and some start with a zero that a number would lose.

## From lists and dictionaries to tables

Two patterns turn up again and again, and both become tables in the next episode:

- a **dictionary of lists**: each key is a column, and each list holds that column's values
- a **list of dictionaries**: each dictionary is one row (one record)

````python
# dictionary of lists: one key per column
table_by_column = {
    "name": ["Marion", "Lake", "Allen"],
    "population": [918977, 493618, 360412],
}

# list of dictionaries: one dictionary per row
table_by_row = [
    {"name": "Marion", "population": 918977},
    {"name": "Lake", "population": 493618},
    {"name": "Allen", "population": 360412},
]
print(table_by_column["name"][1])     # column "name", second row
print(table_by_row[1]["name"])        # second row, column "name"
````

## Recap

- Lists keep values in order; indexes start at 0 and `[-1]` is the last item
- Slices `[start:end]` include `start` but not `end`
- Dictionaries store `key: value` pairs; look values up by key
- Store identifiers such as FIPS codes as text
- A dictionary of lists and a list of dictionaries are both ways to describe a table
