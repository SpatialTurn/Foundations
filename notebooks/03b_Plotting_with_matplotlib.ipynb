**matplotlib** is Python's most widely used plotting library. Its `pyplot` part, nicknamed `plt`, has one
function per chart type. We use the same Indiana county table as notebook 03a.

````python
import pandas as pd
import matplotlib.pyplot as plt

counties = pd.read_csv("https://raw.githubusercontent.com/SpatialTurn/Foundations/main/data/indiana_county_stats.csv")
counties["density"] = counties["pop_2012"] / counties["land_sq_mi"]
````

## Bar chart: comparing a few categories

````python
top = counties.sort_values("pop_2012", ascending=False).head(10)

plt.figure(figsize=(8, 4))
plt.bar(top["name"], top["pop_2012"], color="steelblue")
plt.title("The ten most populous Indiana counties, 2012")
plt.ylabel("Population")
plt.xticks(rotation=45, ha="right")    # tilt long labels so they do not overlap
plt.show()
````

Every chart needs a title and axis labels with units. `plt.show()` draws the chart; put it last.

## Histogram: the shape of one variable

A histogram groups values into bins and counts how many fall in each. It shows where most values sit and
whether there are outliers.

````python
plt.hist(counties["unemp_2016"], bins=12, color="steelblue", edgecolor="white")
plt.title("Unemployment across Indiana's 92 counties, 2016")
plt.xlabel("Unemployment rate (%)")
plt.ylabel("Number of counties")
plt.show()
````

## Scatter plot: the relationship between two variables

Population density is very skewed: Marion County is more than ten times as dense as most counties. A
**logarithmic** x axis spreads the values out so the pattern is visible.

````python
plt.scatter(counties["density"], counties["unemp_2016"], alpha=0.7)
plt.xscale("log")
plt.title("Density and unemployment, Indiana counties")
plt.xlabel("People per square mile (log scale)")
plt.ylabel("Unemployment rate, 2016 (%)")
plt.show()
````

### Marker size as a third variable

The `s=` parameter sets each marker's area. Large numbers such as populations must be scaled down first,
or the markers cover the whole chart.

````python
plt.scatter(counties["density"], counties["unemp_2016"],
            s=counties["pop_2012"] / 2000, alpha=0.5)
plt.xscale("log")
plt.title("Marker size shows population")
plt.xlabel("People per square mile (log scale)")
plt.ylabel("Unemployment rate, 2016 (%)")
plt.show()
````

## Which chart?

| Question | Chart | matplotlib |
|---|---|---|
| How do a few categories compare? | Bar chart | `plt.bar()` or `plt.barh()` |
| How are values of one variable spread out? | Histogram | `plt.hist()` |
| Are two variables related? | Scatter plot | `plt.scatter()` |
| How does a value change over time? | Line chart | `plt.plot()` |

The Data Visualization module covers chart design in depth.

## Several charts in one figure, and saving

`plt.subplots()` makes a figure with several panels (axes). `savefig()` writes the figure to a file.

````python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].hist(counties["unemp_2016"], bins=12, color="steelblue", edgecolor="white")
axes[0].set_title("Unemployment rate (%)")
axes[1].hist(counties["pop_2012"] / 1000, bins=20, color="darkorange", edgecolor="white")
axes[1].set_title("Population (thousands)")
fig.suptitle("Indiana counties")
fig.savefig("indiana_histograms.png", dpi=150, bbox_inches="tight")
plt.show()
````

pandas has a shortcut that calls matplotlib for you: `counties.plot.scatter(x="density", y="unemp_2016")`.
