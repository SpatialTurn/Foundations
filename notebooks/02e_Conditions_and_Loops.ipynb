## Comparisons give True or False

| Operator | Meaning |
|---|---|
| `==` | equal to (two equals signs; one `=` stores a value) |
| `!=` | not equal to |
| `>` `<` | greater than, less than |
| `>=` `<=` | greater than or equal to, less than or equal to |

````python
density = 355.2
print(density > 300)
print(density == 300)
print("Tippecanoe" != "Marion")
````

## if, elif and else

An `if` block runs only when its condition is `True`. The lines that belong to it are **indented** (four
spaces); the indentation is how Python knows where the block ends. `elif` ("else if") checks another
condition, and `else` catches everything left over.

````python
density = 355.2

if density >= 1000:
    label = "very dense"
elif density >= 100:
    label = "moderately dense"
else:
    label = "rural"

print(f"{density} people per square mile is {label}")
````

Change `density` to `50` or `1500` and run the cell again.

### Combining conditions

````python
unemployment = 5.4
population = 12502

if unemployment > 5 and population < 50000:
    print("A small county with high unemployment")
if unemployment > 6 or population > 500000:
    print("This line does not print")
print(not unemployment > 5)
````

## for loops

A `for` loop runs the same block once for each item in a list. The loop variable (`name` below) takes
each value in turn.

````python
counties = ["Marion", "Lake", "Allen"]
for name in counties:
    print(f"{name} County")
````

### A loop with a decision inside

This is the pattern behind a lot of data cleaning and classification.

````python
densities = [101.4, 548.3, 194.5, 21.7, 75.7]
for d in densities:
    if d >= 300:
        print(d, "urban")
    else:
        print(d, "rural or small town")
````

### Building up a result

Start with an empty value before the loop, and add to it inside the loop.

````python
populations = [34365, 360412, 79129, 8804, 12502]
total = 0
big = []
for p in populations:
    total += p
    if p > 50000:
        big.append(p)
print("Total:", total)
print("Counties over 50,000:", big)
````

### Useful helpers

- `range(5)` gives the numbers 0 to 4, handy for repeating something a set number of times.
- `for key, value in my_dict.items():` loops over a dictionary.
- A **list comprehension** builds a new list in one line: `[d * 2.59 for d in areas]`.

````python
for i in range(3):
    print("Round", i)

county = {"name": "Allen", "population": 360412, "unemployment": 4.3}
for key, value in county.items():
    print(key, "=", value)

areas_sq_mi = [339.0, 657.3, 406.9]
areas_km2 = [round(a * 2.59, 1) for a in areas_sq_mi]
print(areas_km2)
````

A `while` loop repeats as long as a condition is true. Make sure something inside the loop eventually makes
the condition false, or the loop never stops (if that happens, click the stop button next to the cell).

````python
count = 0
while count < 3:
    print("count is", count)
    count += 1
````
