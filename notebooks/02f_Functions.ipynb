## Why functions?

You have already used functions: `print()`, `len()`, `round()`, `sum()`. Writing your own lets you give a
name to a piece of work and reuse it, instead of copying the same lines again and again. If you find a
mistake, you fix it in one place.

## Defining a function

````python
def density(population, area):
    """Return people per unit of area."""
    return population / area

print(density(177513, 499.8))
print(density(34365, 339.0))
````

- `def` starts the definition, followed by the function's name and its **parameters** in brackets.
- The indented lines are the body. The text in triple quotes is a **docstring**: a short description.
- `return` sends a result back to whoever called the function. The function stops at `return`.

### return is not print

`print()` only shows a value. `return` hands it back so you can store it and use it again.

````python
def area_km2(area_sq_mi):
    return area_sq_mi * 2.59

tippecanoe_km2 = area_km2(499.8)        # the returned value is stored
print(round(tippecanoe_km2, 1))
print(round(density(177513, tippecanoe_km2), 1), "people per square kilometer")
````

## Default values and keyword arguments

A parameter can have a **default**, used when the caller does not give a value. Callers can also name the
arguments, which makes calls easier to read.

````python
def describe(name, population, unit="people"):
    return f"{name}: {population:,} {unit}"

print(describe("Tippecanoe", 177513))
print(describe("Tippecanoe", 68000, unit="households"))
print(describe(population=8804, name="Benton"))
````

## Functions that make decisions

Functions combine nicely with `if` blocks and loops.

````python
def density_class(d):
    """Label a population density (people per square mile)."""
    if d >= 1000:
        return "very dense"
    elif d >= 100:
        return "moderately dense"
    return "rural"

for d in [101.4, 2318.9, 21.7]:
    print(d, density_class(d))
````

## Variables inside a function stay inside

A variable created inside a function exists only while the function runs. Pass values in through
parameters and get results out with `return`, rather than relying on variables from elsewhere.

````python
def double(x):
    result = x * 2
    return result

print(double(21))
````

````python
print(result)   # result only existed inside double()
````
